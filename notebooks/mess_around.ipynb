```python
import pandas as pd

fd_001_processed = pd.read_csv('../data/processed/FD001_processed.csv')
fd_001_processed.head()
fd_001_processed.describe()
```

```python
import numpy as np
true_rul = pd.read_csv('../data/CMAPSSData/RUL_FD001.txt', sep=r'\s+', header=None)
true_rul = np.array(true_rul[0])
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error

mse = mean_squared_error(true_rul, predictions)
print(f'Root Mean Squared Error: {mse ** 0.5}')
mae = mean_absolute_error(true_rul, predictions)
print(f'Mean Absolute Error: {mae}')
```

Lin reg performs poorly. Could be because
1. Engine degradation is not linear (i.e. sensor value changes don't always lead to changes like in the coefficients. There might be dependency between the sensors, plus you would expect changes in temp to indicate different things for an engine that has run for 100 cycles vs. 20 cycles.)
2. Could be that different sensor measurements together say something too as in temp high, pressure low = near failure. Lin reg can't capture that properly. Decision tree would be better at this, as it can literally do if x > 100, y < 20, we close to failure. 
3. We're only using the current snapshot for prediction i.e. the last cycle measurement. You would expect prev cycles to contain info about the degradation too!

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(max_depth=5, random_state=42) #max_depth prevents overfitting (otherwise tree can grow too deep and memorize the training data), random state ensures reproducibility 
```

```python
tree_reg.fit(X, Y)
```

```python
d_tree_preds = tree_reg.predict(test_df)
d_tree_preds
```

```python
d_tree_mse = mean_squared_error(true_rul, d_tree_preds)
print(f'Decision Tree Root Mean Squared Error: {d_tree_mse ** 0.5}')
d_tree_mae = mean_absolute_error(true_rul, d_tree_preds)
print(f'Decision Tree Mean Absolute Error: {d_tree_mae}')
```

random forest doesn't seem to work much better either. Could be because
1. Most of the data is actually linear. this is failure mode 1, so we expect data to be simple..
2. We are not using the whole trajectory, means that each model only has information about last cycle, might not be enough for both lin reg and dtree reg. 
3. Underfitting doesn't seem to be issue. Increasing tree depth made the error worse. 
4. Maybe need to use many trees

```python
from sklearn.ensemble import RandomForestRegressor

rf_reg = RandomForestRegressor(n_estimators=100, max_depth=8, random_state=42) #n_estimators is the number of trees in the forest

#n_estimators = number of trees in the forest, max_depth = maximum depth of each tree, random_state = seed for reproducibility
#more estimators = more trees = better performance but slower training and prediction time, max_depth = bias variance trade off, deeper trees = lower bias but higher variance, shallower trees = higher bias but lower variance
```

```python
rf_reg.fit(X, Y)
```

```python
rf_preds = rf_reg.predict(test_df)
```

```python
rf_reg_mse = mean_squared_error(true_rul, rf_preds)
print(f'Random Forest Root Mean Squared Error: {rf_reg_mse ** 0.5}')
```

```python
rf_reg_mae = mean_absolute_error(true_rul, rf_preds)
print(f'Random Forest Mean Absolute Error: {rf_reg_mae}')
```

#RF seems to do slightly better, but not that much better. I think the problem is that we're using a snapshot of the last cycle rather than giving it all the data of the trajectory. i.e. non-linearity doesn't seem to be the only challenge.
